#**Functions**

# Functions
A function is a block of code designed to perform a specific task.

```python
# Define the funtion
def add_three(input_var):
  output_var = input_var + 3
  return output_var
```

Every function is composed of two pieces: a *header* and a *body*.

The function **header** defines the name of the function and its argument(s).
*   Every function header begins with 'def' followed by the function name.
*   The **argument** is the name of the variable that will be used as input to the function. It is enclosed in parentheses that appear after the name of the function.
*   For every function, the parentheses enclosing the function argument(s) must be followed by a colon (:).

The function **body** specifies the work that the function does.
*   Every line of code in the function body must be indented exactly four spaces.

A **return statement**  returns the value of a functon's output.

A function is *called* to run.


```python
# Run the function with 10 as input
new_number = add_three(10)

# Check that the value is 13, as expected
print(new_number)
```

When a function is referred in other parts of the code, empty closing parentheses is used to indicate a function as opposed to a variable or another Python object. These parentheses should always be empty, even if the function has arguments.

When naming a function, only lowercase letters should be used, with words separated by underscores.

```python
def get_pay(num_hours):
  # Pre-tax pay, based on receiving $15/hour
  pay_pretax = num_hours * 15
  # After-tax pay, based on being in 12% tax bracket
  pay_aftertax = pay_pretax * (1 - .12)
  return pay_aftertax
```

```python
# Calculate pay based on working 40 hours
pay_fulltime = get_pay(40)
print(pay_fulltime)
```

```python
pay_parttime = get_pay(32)
print(pay_parttime)
```

Variables defined inside a function body cannot be accessed outside of the function.

```python
print(pay_aftertax)
```

A variable's **scope** is the part of the code where it is accessible. Variables defined inside a function have a *local scope* and variables defined outside a function have a *global scope*.

More arguments can be added inside the parentheses in a function head, separated with a comma, to create a function with multiple arguments. To call such a function, values should be assigned for each input.

```python
def get_pay_with_more_inputs(num_hours, hourly_wage, tax_bracket):
  # Pre-tax pay
  pay_pretax =  num_hours * hourly_wage
  # After-tax pay
  pay_aftertax = pay_pretax * (1 - tax_bracket)
  return pay_aftertax
```

```python
higher_pay_aftertax = get_pay_with_more_inputs(40, 24, .22)
print(higher_pay_aftertax)
```

```python
same_pay_fulltime = get_pay_with_more_inputs(40, 15, .12)
print(same_pay_fulltime)
```

```python
# Define the function with no arguments and with no return
def print_hello():
  print("Hello, you!")
  print("Good morning!")

# call the function
print_hello()
```

# Conditions
Conditions are statements that are either 'True' or 'False'.

```python
print(2 > 3)
```

```python
var_one = 1
var_two = 2

print(var_one < 1)
print(var_two >= var_one)
```

Common symbols useed to construct conditions include:
*   (==) equals
*   (!=) does not equal
*   (<) less than
*   (<=) less than or equal to
*   (>) greater than
*   (>=) greater than or equal to

# Conditional Statements
Conditional statements use conditions to modify how a function runs. They run a certain block of code if the value of a condition evaluates to 'True'.

The simplest type of conditional statement is an 'if' statement

```python
def evaluate_temp(temp):
  # Set an initial message
  message = "Normal temperature"
  # Update value of message only if temperature greater then 38
  if temp > 38:
    message = "Fever!"
  return message
```

```python
print(evaluate_temp(37))
```

```python
print(evaluate_temp(39))
```

An 'else' statement runs code if an 'if' statement is False.

```python
def evaluate_temp_with_else(temp):
  if temp > 38:
    message = "Fever!"
  else:
    message = "Normal temperature."
  return message
```

```python
print(evaluate_temp_with_else(37))
```

An 'elif' (short for 'else if') statement checks if multiple conditions might be true.

```python
def evaluate_temp_with_elif(temp):
  if temp > 38:
    message = "Fever!"
  elif temp > 35:
    message = "Normal temperature."
  else:
    message = "Low temperature."
  return message
```

```python
evaluate_temp_with_elif(36)
```

```python
evaluate_temp_with_elif(34)
```

```python
def get_dose(weight):
  # Dosage is 1.25 ml for anyone uder 5.2 kg
  if weight < 5.2:
    dose = 1.25
  elif weight < 7.9:
    dose = 2.5
  elif weight < 10.4:
    dose = 3.75
  elif weight < 15.9:
    dose = 5
  elif weight < 21.2:
    dose = 7.5
  # Dosage is 10 ml for anyone 21.2 kg or over
  else:
    dose = 10
  return dose
```

# **Intro to lists**


# Lists
Strings can be organized using a list. To create a list, square brackets ([]) are used and each item within are separated with a comma.


```python
flowers = "pink primrose, hard-leaved pocket orchid, canterbury bells, sweet pea, english marigold, tiger lily, moon orchid, bird of paradise, monkshood, globe thistle"

print(type(flowers))
print(flowers)
```

```python
flowers_list = ["pink primrose", "hard-leaved pocket orchid", "canterbury bells", "sweet pea", "english marigold", "tiger lily", "moon orchid", "bird of paradise", "monkshood", "globe thistle"]

print(type(flowers_list))
print(flowers_list)
```

The number of **entries** in any list can be counted with 'len()'

```python
print(len(flowers_list))
```

Any item in the list can be referred according to its position. This is called **indexing**. Python uses zero-based indexing, which means that
*   the first item is indexed as 0.
*   the last item is one less than the length of the list.

```python
print("First entry:", flowers_list[0])
print("Second entry:", flowers_list[1])

# The list has length ten, so we refer to final entry with 9
print("Last entry:", flowers_list[9])
```

Segments of a list can be pulled. This is called **slicing**.
*   To pull the first x entries, [:x] is used, and
*   To pull the last y entries, [-y:] is used

Slicing returns a shortened list.

```python
print("First three entries:", flowers_list[:3])
print("Final two entries:", flowers_list[-2:])
```

Items can be removed from a list using '.remove()'.

```python
flowers_list.remove("globe thistle")
print(flowers_list)
```

Items can be added to a list using '.append()'.

```python
flowers_list.append("snapdragon")
print(flowers_list)
```

Lists, not only can contain strings, but can have any data types.

```python
hardcover_sales = [139, 128, 172, 139, 191, 168, 170]
```

```python
print("Length of the list:", len(hardcover_sales))
print("Entry at index 2:", hardcover_sales[2])
```

The minimum and the maximum can be obtained using 'min()' and 'max()' respectively.

```python
print("Minimum:", min(hardcover_sales))
print("Maximum:", max(hardcover_sales))
```

'sum()' adds every item in the list.

```python
print("Total books sold in one week:", sum(hardcover_sales))
```

Similar calculations can be done with slices.

```python
print("Average books sold in first five days:", sum(hardcover_sales[:5]) / 5)
```
